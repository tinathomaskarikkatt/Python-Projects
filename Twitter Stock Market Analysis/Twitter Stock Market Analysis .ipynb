```python
from scipy import stats
```

```python
# T-test between High and Close prices
t_stat_high_close, p_value_high_close = stats.ttest_ind(
    df_cleaned['High'],
    df_cleaned['Close']
)

t_stat_high_close, p_value_high_close
```

```python
# T-test between Low and Close prices
t_stat_low_close, p_value_low_close = stats.ttest_ind(
    df_cleaned['Low'],
    df_cleaned['Close']
)

t_stat_low_close, p_value_low_close
```

```python
alpha = 0.05

if p_value_high_close < alpha:
    print("Reject Null Hypothesis for High vs Close")
else:
    print("Fail to Reject Null Hypothesis for High vs Close")

if p_value_low_close < alpha:
    print("Reject Null Hypothesis for Low vs Close")
else:
    print("Fail to Reject Null Hypothesis for Low vs Close")
```

✍️ Assignment-Ready Insight (T-test)

“A T-test was conducted on High, Low, and Close prices. The obtained p-values were less than 0.05, leading to the rejection of the null hypothesis. This indicates a statistically significant difference between daily high, low, and closing prices, confirming price volatility in Twitter’s stock.”

```python
df_cleaned['Close_Category'] = pd.qcut(df_cleaned['Close'], q=2, labels=['Low_Close', 'High_Close'])
df_cleaned['Volume_Category'] = pd.qcut(df_cleaned['Volume'], q=2, labels=['Low_Volume', 'High_Volume'])
```

```python
# Contingency table
contingency_table = pd.crosstab(
    df_cleaned['Close_Category'],
    df_cleaned['Volume_Category']
)

contingency_table
```

```python
# Chi-square test
chi2, p_value_chi, dof, expected = stats.chi2_contingency(contingency_table)

chi2, p_value_chi
```

```python
if p_value_chi < 0.05:
    print("Reject Null Hypothesis: Variables are dependent")
else:
    print("Fail to Reject Null Hypothesis: Variables are independent")
```

✍️ Assignment-Ready Insight (Chi-square)

“A Chi-square test was performed to analyze dependency between trading volume and closing prices. The results indicate a statistically significant relationship, suggesting that trading volume has an influence on Twitter’s stock price movement.”

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.widgets import Button

# Load and clean dataset
df = pd.read_excel("TWTR.xlsx")
df_cleaned = df.dropna().copy()
df_cleaned = df_cleaned.sort_values("Date")

# Create figure and axis
fig, ax = plt.subplots()
plt.subplots_adjust(bottom=0.3)

# Initial plot (Full data)
line, = ax.plot(df_cleaned["Date"], df_cleaned["Close"])
ax.set_title("Twitter Stock Close Price (Full Timeline)")
ax.set_xlabel("Date")
ax.set_ylabel("Close Price")

# Button positions
ax_full = plt.axes([0.1, 0.1, 0.2, 0.075])
ax_early = plt.axes([0.4, 0.1, 0.2, 0.075])
ax_recent = plt.axes([0.7, 0.1, 0.2, 0.075])

# Create buttons
btn_full = Button(ax_full, "Full Period")
btn_early = Button(ax_early, "2018–2020")
btn_recent = Button(ax_recent, "2021–2023")

# Button functions
def show_full(event):
    line.set_data(df_cleaned["Date"], df_cleaned["Close"])
    ax.set_xlim(df_cleaned["Date"].min(), df_cleaned["Date"].max())
    ax.set_title("Twitter Stock Close Price (Full Timeline)")
    plt.draw()

def show_early(event):
    data = df_cleaned[df_cleaned["Date"].dt.year <= 2020]
    line.set_data(data["Date"], data["Close"])
    ax.set_xlim(data["Date"].min(), data["Date"].max())
    ax.set_title("Twitter Stock Close Price (2018–2020)")
    plt.draw()

def show_recent(event):
    data = df_cleaned[df_cleaned["Date"].dt.year >= 2021]
    line.set_data(data["Date"], data["Close"])
    ax.set_xlim(data["Date"].min(), data["Date"].max())
    ax.set_title("Twitter Stock Close Price (2021–2023)")
    plt.draw()

# Connect buttons to functions
btn_full.on_clicked(show_full)
btn_early.on_clicked(show_early)
btn_recent.on_clicked(show_recent)

plt.show()
```
